```python
# 3 Code 1 — Greedy Algorithm using Arithmetic Mean
# Greedy Algorithm using Arithmetic Mean


n = int(input("Enter number of machines: "))
machines = {}


for i in range(n):
    name = input("Enter machine name: ")
    rewards = list(map(int, input("Enter rewards separated by space: ").split()))
    machines[name] = rewards


q_values = {}


for name in machines:
    rewards = machines[name]
    q = sum(rewards) / len(rewards)
    q_values[name] = q


    print("\nMachine", name)
    print("Rewards:", rewards)
    print("Q(" + name + ") =", round(q, 2))


# Find machine with highest Q value
selected = max(q_values, key=q_values.get)


print("\nGreedy Selected Machine:", selected)
```

```python
# Code 2 — Greedy Algorithm using Incremental Update
# Greedy Algorithm using Incremental Update


machines = {
    "A": [3, 2, 4],
    "B": [6, 7],
    "C": [5, 4, 6]
}


Q = {"A": 0, "B": 0, "C": 0}
N = {"A": 0, "B": 0, "C": 0}


for machine in machines:
    print("\nMachine", machine)


    for reward in machines[machine]:
        N[machine] += 1


        # Incremental update
        Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]


        print("Reward:", reward)
        print("N =", N[machine], "Q =", round(Q[machine], 2))


# Find the machine with highest Q value
selected = max(Q, key=Q.get)


print("\nFinal Q Values:")
for machine in Q:
    print("Q(" + machine + ") =", round(Q[machine], 2))


print("\nGreedy Selected Machine:", selected)
```
